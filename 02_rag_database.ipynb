```python
import os
from langchain_community.document_loaders import TextLoader

# 👇 THIS IS THE UPDATED LINE 👇
from langchain_text_splitters import RecursiveCharacterTextSplitter 

from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

print("Step 1: Loading the biological documents...")
loader = TextLoader("./knowledge_base/genetics_info.txt")
documents = loader.load()

print("Step 2: Splitting text into manageable chunks...")
text_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=50)
docs = text_splitter.split_documents(documents)
print(f"Split document into {len(docs)} chunks.")

print("Step 3: Initializing local embedding model...")
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

print("Step 4: Building the ChromaDB Vector Database...")
vectorstore = Chroma.from_documents(
    documents=docs, 
    embedding=embeddings, 
    persist_directory="./chroma_db"
)

print("\n✅ Knowledge Base built successfully! Your AI can now read biology.")

# Let's test if the search engine works!
query = "What letters are in the Exon-Intron donor site?"
results = vectorstore.similarity_search(query, k=1)
print(f"\nTest Search Result: {results[0].page_content}")
```
